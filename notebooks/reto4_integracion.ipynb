```python
import pandas as pd

temp = pd.read_csv("resultado_temperatura.csv")
ph = pd.read_csv("resultado_ph.csv")
ox = pd.read_csv("resultado_oxigeno.csv")
```

```python
temp.head()
```

```python
ph.head()
```

```python
ox.head()
```

```python
df_final = pd.concat([temp, ph, ox], ignore_index=True)

df_final
```

```python
df_final.to_csv("resultado_integrado.csv", index=False)

print("Archivo integrado creado correctamente")
```
