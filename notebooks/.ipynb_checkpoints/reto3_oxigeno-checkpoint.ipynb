# 🧩 Reto 3 – Oxígeno disuelto
**Objetivo:** evaluar riesgo ecológico según oxígeno disuelto.

**Datos:** `../data/muestras_calidad_agua.csv`

### Pasos
1. Carga el CSV en `df`.
2. Extrae `sitio` y `oxigeno`.
3. Crea `nivel_oxigeno` (bajo/medio/alto).
4. Crea `riesgo_biologico` (⚠ si oxígeno < 5).
5. Filtra y exporta `resultado_oxigeno.csv`.

> **Reto adicional:** porcentaje de sitios con riesgo.

```python
import pandas as pd
import numpy as np

# 1. Carga con manejo de errores
try:
    df = pd.read_csv("../data/muestras_calidad_agua.csv")
except FileNotFoundError:
    print("Error: No se encontró el archivo CSV en la ruta especificada.")
    exit()

# Limpiar nombres de columnas (espacios y minúsculas para facilitar búsqueda)
df.columns = df.columns.str.strip().str.lower()

# 2. Identificación dinámica de columnas (más flexible)
def encontrar_columna(nombres_posibles, columnas_df):
    for col in columnas_df:
        if any(opcion in col for opcion in nombres_posibles):
            return col
    return None

col_sitio = encontrar_columna(["sitio", "estacion"], df.columns)
col_oxigeno = encontrar_columna(["oxigeno", "oxígeno", "do", "o2"], df.columns)

if not col_sitio or not col_oxigeno:
    raise ValueError("No se encontraron las columnas necesarias en el CSV.")

# Crear el DataFrame de trabajo seleccionando solo lo necesario
df_oxigeno = df[[col_sitio, col_oxigeno]].copy()
df_oxigeno.columns = ["sitio", "oxigeno"]

# Asegurar que 'oxigeno' sea numérico (por si hay datos corruptos o strings)
df_oxigeno["oxigeno"] = pd.to_numeric(df_oxigeno["oxigeno"], errors='coerce')
df_oxigeno = df_oxigeno.dropna(subset=["oxigeno"]) # Eliminar filas sin datos de oxígeno

# 3 y 4. Clasificación usando np.select (Más limpio que nested np.where)
condiciones = [
    (df_oxigeno["oxigeno"] < 5),
    (df_oxigeno["oxigeno"] >= 5) & (df_oxigeno["oxigeno"] < 7)
]
opciones = ["Bajo", "Medio"]

df_oxigeno["clasificacion_oxigeno"] = np.select(condiciones, opciones, default="Bueno")

# Crear estres_biológico de forma booleana (más eficiente para filtros)
df_oxigeno["estres_biologico"] = np.where(df_oxigeno["oxigeno"] < 5, "Sí", "No")

# 5. Exportar y mostrar métricas
df_oxigeno.to_csv("resultado_oxigeno.csv", index=False)

print(f"--- Resumen de Calidad ---")
print(f"Sitios analizados: {len(df_oxigeno)}")
print(f"Promedio de oxígeno: {df_oxigeno['oxigeno'].mean():.2f} mg/L")
print(f"Valor mínimo: {df_oxigeno['oxigeno'].min():.2f} mg/L")
print(f"Sitios con estrés biológico: {len(df_oxigeno[df_oxigeno['estres_biologico'] == 'Sí'])}")
```

```python
import pandas as pd
import numpy as np

# 1. Cargar el CSV en df
df = pd.read_csv("../data/muestras_calidad_agua.csv")

# Limpiar nombres de columnas
df.columns = df.columns.str.strip()

# 2. Extraer sitio y oxígeno
df_oxigeno = df[["sitio", "oxigeno"]].copy()

# 3. Crear clasificacion_oxigeno con tres rangos
df_oxigeno["clasificacion_oxigeno"] = np.where(
    df_oxigeno["oxigeno"] < 5,
    "Bajo",
    np.where(
        df_oxigeno["oxigeno"] < 7,
        "Medio",
        "Bueno"
    )
)

# 4. Crear estres_biologico
df_oxigeno["estres_biologico"] = np.where(
    df_oxigeno["oxigeno"] < 5,
    "Sí",
    "No"
)

# 5. Filtrar sitios con estrés biológico
df_estres = df_oxigeno[df_oxigeno["estres_biologico"] == "Sí"]

# Exportar resultado
df_oxigeno.to_csv("resultado_oxigeno.csv", index=False)

# Reto adicional: promedio y valor mínimo
promedio_oxigeno = df_oxigeno["oxigeno"].mean()
minimo_oxigeno = df_oxigeno["oxigeno"].min()

print("Archivo resultado_oxigeno.csv creado correctamente.")
print("Promedio de oxígeno:", round(promedio_oxigeno, 2), "mg/L")
print("Valor mínimo de oxígeno:", minimo_oxigeno, "mg/L")

df_oxigeno
```
