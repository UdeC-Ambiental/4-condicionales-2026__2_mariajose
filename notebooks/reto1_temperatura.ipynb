# 🧩 Reto 1 – Temperatura
**Objetivo:** clasificar la temperatura del agua y evaluar posible estrés térmico.

**Datos:** `../data/muestras_calidad_agua.csv`

### Pasos
1. Carga el CSV en `df`.
2. Extrae `sitio` y `temperatura` en `df_temp`.
3. Crea `categoria_temp` con np.where o .loc.
4. Crea `riesgo_termico`.
5. Filtra a `df_caliente` (alta temperatura).
6. Exporta `resultado_temperatura.csv`.

> **Reto adicional:** porcentaje de sitios con riesgo térmico.

```python
import pandas as pd
import numpy as np

# 1. Cargar el CSV en df
df = pd.read_csv("../data/muestras_calidad_agua.csv")


# Verificar columnas disponibles
df.head()
```

```python
# Limpiar posibles espacios en los nombres de columnas
df.columns = df.columns.str.strip()

# Buscar automáticamente las columnas necesarias
col_sitio = [col for col in df.columns if "sitio" in col.lower()][0]
col_temperatura = [col for col in df.columns if "temperatura" in col.lower()][0]

# 2. Extraer sitio y temperatura en df_temp
df_temp = df[[col_sitio, col_temperatura]].copy()

# Renombrar columnas para trabajar más fácil
df_temp.columns = ["sitio", "temperatura"]

df_temp
```

```python
# 3. Crear categoria_temp usando np.where

df_temp["categoria_temp"] = np.where(
    df_temp["temperatura"] <= 20,
    "Baja",
    np.where(
        df_temp["temperatura"] <= 26,
        "Media",
        "Alta"
    )
)

df_temp
```

```python
# 4. Crear riesgo_termico usando np.where

df_temp["riesgo_termico"] = np.where(
    df_temp["temperatura"] > 26,
    "Sí",
    "No"
)

df_temp
```

```python
# 5. Filtrar sitios con alta temperatura

df_caliente = df_temp[df_temp["categoria_temp"] == "Alta"]

df_caliente
```

```python
# 6. Exportar resultado a un archivo CSV

df_temp.to_csv("resultado_temperatura.csv", index=False)

print("Archivo resultado_temperatura.csv creado correctamente.")
```

```python
# Reto adicional: porcentaje de sitios con riesgo térmico

total_sitios = len(df_temp)
sitios_con_riesgo = len(df_temp[df_temp["riesgo_termico"] == "Sí"])

porcentaje_riesgo = (sitios_con_riesgo / total_sitios) * 100

print("Total de sitios:", total_sitios)
print("Sitios con riesgo térmico:", sitios_con_riesgo)
print("Porcentaje de sitios con riesgo térmico:", round(porcentaje_riesgo, 2), "%")
```

## Conclusión

A partir del análisis de la temperatura, se clasificaron los sitios en temperatura baja, media y alta.

Los sitios con temperatura alta se consideran con posible riesgo térmico, ya que las temperaturas elevadas pueden reducir la cantidad de oxígeno disponible en el agua y afectar la vida acuática.

El porcentaje calculado permite identificar qué proporción de los sitios evaluados presenta riesgo térmico.
