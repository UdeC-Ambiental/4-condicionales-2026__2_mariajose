# 🧩 Reto 2 – pH
**Objetivo:** clasificar pH y generar alerta química.

**Datos:** `../data/muestras_calidad_agua.csv`

### Pasos
1. Carga el CSV en `df`.
2. Extrae `sitio` y `pH`.
3. Crea `clasificacion_pH` con tres rangos.
4. Crea `alerta_quimica` (revisión si pH < 6.5 o pH > 8.5).
5. Filtra `df_alerta` y exporta `resultado_ph.csv`.

> **Reto adicional:** promedio y desviación estándar.

```python
import pandas as pd
import numpy as np

# 1. Cargar el CSV en df
df = pd.read_csv("../data/muestras_calidad_agua.csv")

# Mostrar las primeras filas para verificar los datos
df.head()
```

```python
# Limpiar posibles espacios en los nombres de columnas
df.columns = df.columns.str.strip()

# Buscar automáticamente las columnas necesarias
col_sitio = [col for col in df.columns if "sitio" in col.lower()][0]
col_ph = [col for col in df.columns if "ph" in col.lower()][0]

# 2. Extraer sitio y pH en df_ph
df_ph = df[[col_sitio, col_ph]].copy()

# Renombrar columnas para trabajar más fácil
df_ph.columns = ["sitio", "pH"]

df_ph
```

```python
# 3. Crear clasificacion_pH con tres rangos

df_ph["clasificacion_pH"] = np.where(
    df_ph["pH"] < 6.5,
    "Ácido",
    np.where(
        df_ph["pH"] <= 8.5,
        "Adecuado",
        "Alcalino"
    )
)

df_ph
```

```python
# 4. Crear alerta_quimica

df_ph["alerta_quimica"] = np.where(
    (df_ph["pH"] < 6.5) | (df_ph["pH"] > 8.5),
    "Sí",
    "No"
)

df_ph
```

```python
# 5. Filtrar sitios con alerta química

df_alerta = df_ph[df_ph["alerta_quimica"] == "Sí"]

df_alerta
```

```python
# Exportar resultado a un archivo CSV

df_ph.to_csv("resultado_ph.csv", index=False)

print("Archivo resultado_ph.csv creado correctamente.")
```

```python
# Reto adicional: promedio y desviación estándar del pH

promedio_ph = df_ph["pH"].mean()
desviacion_ph = df_ph["pH"].std()

print("Promedio de pH:", round(promedio_ph, 2))
print("Desviación estándar del pH:", round(desviacion_ph, 2))
```

## Conclusión

A partir del análisis del pH, se clasificaron los sitios en tres categorías: ácido, adecuado y alcalino.

Los valores de pH menores a 6.5 se consideran ácidos, mientras que los valores mayores a 8.5 se consideran alcalinos. Estos casos generan una alerta química porque están fuera del rango adecuado.

El promedio y la desviación estándar permiten observar el comportamiento general del pH en los sitios analizados.
